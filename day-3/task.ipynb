```python
%sql
-- 3
DESCRIBE cyntexa_dev.sales.customer
```

```python
%sql
DESCRIBE EXTENDED cyntexa_dev.sales.customer
```

```python
%sql
DESCRIBE DETAIL cyntexa_dev.sales.customer
```


- DESCRIBE - it only shows the column and it s datatype with comments of a table

- DESCRIBE EXTENDED - 
It shows the details of catalog,schema,location,table type like managed,external,created_date .

- DESCRIBE DETAIL - 
It shows the table details like number of parquet files present , and more detail.

#INTERMEDIATE

```python
%sql
-- 4 - Add metadata
SELECT *,_metadata.file_name,_metadata.file_path
FROM cyntexa_dev.sales.customer
```

```python
%sql
-- 5 -- ICEBERG TABLE
CREATE TABLE cyntexa_dev.sales.customer_iceberg  USING ICEBERG AS
SELECT * FROM cyntexa_dev.sales.customer
```

```python
%sql
DESCRIBE DETAIL cyntexa_dev.sales.customer_iceberg
```

```python
%sql
-- 6
-- RECORD PER SOURCE FILE

SELECT _metadata.file_name AS source_file , COUNT(*) AS record_count FROM cyntexa_dev.sales.customer 
GROUP BY _metadata.file_name
```
