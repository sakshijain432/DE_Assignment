```python
from pyspark.sql .functions import *

df = spark.read.table("cyntexa_dev.sales.sales_bronze")

clean_df = df.dropDuplicates(["order_id"])
clean_df = clean_df.withColumn("order_date", to_date(col("order_date"), "yyyy-MM-dd"))
clean_df = clean_df.withColumn("total_amount", col("total_amount").cast("double"))
clean_df = clean_df.dropna()

clean_df.write.mode("overwrite").saveAsTable("cyntexa_dev.sales.sales_silver")
```
