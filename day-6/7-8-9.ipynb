```python
%sql
CREATE OR REPLACE TABLE cyntexa_dev.sales.daily_product_inventory AS
SELECT
    order_date,
    product_id,
    SUM(quantity) AS units_sold,
    COUNT(DISTINCT order_id) AS total_orders,
    COUNT(DISTINCT customer_id) AS unique_customers
FROM cyntexa_dev.sales.sales_silver
GROUP BY
    order_date,
    product_id
ORDER BY
    order_date,
    product_id;
```

The second Gold aggregation is designed for the inventory team and summarizes units sold by product and date. It belongs in the Gold layer because it is a business-ready, reusable aggregation rather than raw or cleaned transactional data. Centralizing this logic ensures consistent metrics across teams, avoids repeated ad-hoc queries on the Silver layer, and provides a performant dataset for inventory dashboards and reporting.
