```python
%sql
CREATE OR REPLACE TABLE cyntexa_dev.sales.daily_revenue AS

SELECT
    order_date,
    SUM(total_amount) AS daily_revenue,
    COUNT(DISTINCT order_id) AS total_orders,
    SUM(quantity) AS total_quantity,
    COUNT(DISTINCT customer_id) AS unique_customers

FROM cyntexa_dev.sales.sales_silver
GROUP BY order_date 
ORDER BY order_date;
```
