```python
from pyspark.sql.functions import *

def add_metadata(df) :
    df1 = df.withColumn("ingestion_ts", current_timestamp())
    return df1
    
```

```python
df = spark.read.csv("/Volumes/dev/demo/raw-1000-richest/sales.csv",header=True,inferSchema=True)
df = add_metadata(df)
df.write.mode("overwrite").saveAsTable("cyntexa_dev.sales.sales_bronze")

df.printSchema()
```
