##Genie Agent Curation Checklist

## 1. Schema & Table-Level Metadata
* Table Descriptions: Define the business purpose of the table clearly. State exactly what a single row represents.
* Freshness & Update Frequency: Document how often data is refreshed.
* Primary & Foreign Keys: Explicitly document table relationships and join constraints in the table metadata so Genie builds accurate relational joins.

## 2. Column-Level Metadata

* Column Comments: Label every critical column with a precise business definition. Avoid technical jargon or shorthand .
* Unit of Measure: Explicitly state currencies, time zones, and units.

## 3. Semantic & Executive Alignment

* Business Glossary Mapping: Match technical column names directly to the exact terminology used in executive reports.
* Ambiguity Resolution Rules: Document how to handle overlapping definitions.
* Filter Conditions: Define standard exclusions required for executive reporting.



#8.
Photon's role — makes each individual query run faster
Photon speeds up how a query is executed once it starts running — it's a faster engine underneath, processing the actual data operations (joins, filters, aggregations) quicker than the older engine.

Predictive I/O's role — makes finding the right data faster
Predictive I/O's job is different — it speeds up how quickly Databricks finds and reads only the data it actually needs, by pushing filters early and skipping irrelevant data before it's even decompressed. This matters specifically for selective queries — like a dashboard filter asking for "just this region, just last week." Think of Predictive I/O as a smart GPS — it finds the shortest route to the data instead of scanning everything.

IWM's role — makes sure the right amount of compute shows up instantly
IWM's job is different again — it's not about the query itself at all, it's about resource allocation. It predicts how much compute a query needs and assigns it immediately, without waiting for a slow scale-up process. Think of IWM as an instant staffing manager — the moment a query arrives, it makes sure enough "workers" (compute) are ready right away.

Serverless over classic/pro
Classic only has Photon — so classic still executes fast, but wastes time scanning too much data, and waits for slow compute scaling.

Pro adds Predictive I/O — fixes the data-scanning bottleneck too, but still waits for compute to catch up when demand spikes.

Serverless has all three — no bottleneck is left unsolved: fast execution, fast targeted data reads, and instant compute availability, exactly what unpredictable, need-it-now dashboard queries require.
